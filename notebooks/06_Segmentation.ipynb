```python
import pandas as pd

df = pd.read_parquet('D:/analyst/projects/nyc-taxi-analysis-2025/data/clean_yellow_tripdata_2025-01.parquet')

df['revenue_per_mile'] = df['total_amount'] / df['trip_distance']
```

```python
# Time segments

def time_bucket(hour):
    if 6 <= hour < 10:
        return 'Morning Rush'
    elif 10 <= hour < 16:
        return 'Midday'
    elif 16 <= hour < 20:
        return 'Evening Rush'
    else:
        return 'Night'

df['time_segment'] = df['pickup_hour'].apply(time_bucket)
```

```python
# Distance segments

def distance_bucket(d):
    if d < 2:
        return 'Very Short'
    elif d < 5:
        return 'Short'
    elif d < 10:
        return 'Medium'
    else:
        return 'Long'

df['distance_segment'] = df['trip_distance'].apply(distance_bucket)
```

```python
# Tip segments

def tip_bucket(t):
    if t == 0:
        return 'No Tip'
    elif t < 10:
        return 'Low Tip'
    elif t < 20:
        return 'Normal Tip'
    else:
        return 'High Tip'

df['tip_segment'] = df['tip_pct'].apply(tip_bucket)
```

```python
# Time segments anlaysis

time_analysis = df.groupby('time_segment').agg(
    avg_revenue=('total_amount','mean'),
    avg_distance=('trip_distance','mean'),
    avg_duration=('trip_duration_min','mean'),
    avg_tip=('tip_pct','mean'),
    num_trips=('total_amount','count')
).reset_index()

time_analysis.to_csv('D:/analyst/projects/nyc-taxi-analysis-2025/data/segment_time.csv', index=False)
```

```python
# Distance Segments analysis

distance_analysis = df.groupby('distance_segment').agg(
    avg_revenue=('total_amount','mean'),
    avg_duration=('trip_duration_min','mean'),
    avg_tip=('tip_pct','mean'),
    revenue_per_mile=('revenue_per_mile','mean'),
    num_trips=('total_amount','count')
).reset_index()

distance_analysis.to_csv('D:/analyst/projects/nyc-taxi-analysis-2025/data/segment_distance.csv', index=False)
```

```python
# Tip behavior analysis

tip_analysis = df.groupby('tip_segment').agg(
    avg_revenue=('total_amount','mean'),
    avg_distance=('trip_distance','mean'),
    avg_duration=('trip_duration_min','mean'),
    num_trips=('total_amount','count')
).reset_index()

tip_analysis.to_csv('D:/analyst/projects/nyc-taxi-analysis-2025/data/segment_tip.csv', index=False)
```

```python
# Cross segment (Time and Distance) analysis

time_distance = df.groupby(['time_segment','distance_segment']).agg(
    avg_revenue=('total_amount','mean'),
    avg_tip=('tip_pct','mean'),
    num_trips=('total_amount','count')
).reset_index()

time_distance.to_csv('D:/analyst/projects/nyc-taxi-analysis-2025/data/segment_time_distance.csv', index=False)
```
